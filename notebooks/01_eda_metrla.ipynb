# 01. Exploratory Data Analysis — METR-LA Benchmark

Analyzes the baseline Los Angeles highway sensor network (207 sensors) used as pre-training source for transfer learning.

```python
import numpy as np
import matplotlib.pyplot as plt
import os

print("METR-LA EDA notebook initialized.")
if os.path.exists("notebooks/data/node_values.npy"):
    vals = np.load("notebooks/data/node_values.npy")
    print("Node values shape:", vals.shape)
    plt.figure(figsize=(10, 4))
    plt.plot(vals[:288, 0], label="Sensor 0")
    plt.title("METR-LA 24-hour Speed Profile")
    plt.xlabel("5-min Time Step")
    plt.ylabel("Speed (mph)")
    plt.legend()
    plt.show()
```
