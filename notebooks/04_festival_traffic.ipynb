# 04. Festival Calendar Anomalies & Congestion Peaks

Studies traffic impact across Indian festivals, eve rushes, and commercial corridor saturation.

```python
import pandas as pd
import matplotlib.pyplot as plt

cal = pd.read_csv("data/raw/chennai_festival_calendar_2026.csv")
print(cal.head())

plt.figure(figsize=(8, 4))
cal["traffic_impact"].value_counts().plot(kind="bar", color="coral")
plt.title("Distribution of Expected Traffic Impact Across Festivals")
plt.ylabel("Number of Events")
plt.xlabel("Impact Category")
plt.show()
```
