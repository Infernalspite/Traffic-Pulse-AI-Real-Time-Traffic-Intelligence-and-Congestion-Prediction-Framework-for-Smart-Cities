# 03. Monsoon Rainfall Impact & Waterlogging Risk Analysis

Evaluates speed degradation and capacity reduction during monsoon precipitation events.

```python
import numpy as np
import matplotlib.pyplot as plt

rain_mm_hr = np.linspace(0, 50, 100)
speed_retention = 1.0 - 0.35 * (1.0 - np.exp(-rain_mm_hr / 15.0))

plt.figure(figsize=(8, 4))
plt.plot(rain_mm_hr, speed_retention * 100, color="navy", lw=2)
plt.title("Monsoon Precipitation vs Effective Road Speed Retention")
plt.xlabel("Rainfall Intensity (mm/hr)")
plt.ylabel("Effective Speed Retention (%)")
plt.grid(True, alpha=0.3)
plt.show()
```
