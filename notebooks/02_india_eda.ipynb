# 02. Indian Urban Traffic EDA — Chennai Metropolitan Junctions

Exploratory analysis of 20 major Chennai arterial intersections (Kathipara, Guindy, T. Nagar, Central, OMR).

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("data/raw/chennai_traffic_log_sheet_latest.csv")
print("Total records:", len(df))
print(df.head())

plt.figure(figsize=(12, 5))
df.groupby("junction")["current_speed"].mean().sort_values().plot(kind="barh", color="teal")
plt.title("Average Speed by Chennai Metropolitan Junction")
plt.xlabel("Speed (km/h)")
plt.tight_layout()
plt.show()
```
